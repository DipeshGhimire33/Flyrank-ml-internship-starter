# Proof Statement

I can take a messy, highly imbalanced clinical dataset and build a leakage-free, honestly evaluated prediction model.

My sepsis project:

- **1.55 million ICU records**
- **40,336 patients**
- Splits by **patient rather than by row**
- Selects its **decision threshold on validation data only**
- Benchmarks **XGBoost against four baselines on identical splits**
- Reports a near tie at a **PR-AUC of 0.093** against a **1.8% base rate**, as it is
- Uses **SHAP** to show how heavily the model leans on **ICU stay length**

It is written for a **data lead in healthcare** who needs to trust a junior ML hire's methods before trusting their results.
