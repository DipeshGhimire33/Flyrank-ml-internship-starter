## Case 1: Which Pages Should We Refresh First?

### The Problem

I was handed a dataset and told to pick a lane. Nobody had a refresh process I was replacing, and nobody had labeled which pages actually needed a refresh.

One row per content page: 90 days of impressions, clicks and sessions, a last-30 vs previous-30 comparison, days since the page was last refreshed, and a trend direction and percentage.

The question underneath was simple: **out of all these pages, which ones should a person look at first?**

### What I Did, and What I Decided

I picked ranking over prediction on purpose.

I'm new to machine learning, and I wanted a task I could finish properly with some research and hard work instead of one that sounded impressive and fell apart.

With no label saying "this page needed a refresh," training a model would have meant inventing the answer, so I wrote a scoring rule I could explain instead.

The big decision: **a falling trend outranks an old page.**

A page that hasn't been touched in ages but is still growing can wait. A recently updated page that's dropping is the red flag, because something may be wrong.

I also made sure a big drop on a tiny page doesn't beat a smaller drop on a page people actually see, so pages need real visibility to rank high.

Every page comes out with:

* reason codes: declining performance, stale content, high visibility, ranking opportunity, CTR opportunity, low data confidence
* a confidence note
* a **"what would make this wrong"** column

The score is decision support. It does not promise a refresh will work.

### What Came of It

A ranked list of pages, each with its reasons attached.

In the top 20, every page is flagged as declining and highly visible, which matches the rule I set.

Scores run from about **0.98 down to 0.90**, and all carry medium confidence.

### What I Haven't Done

I didn't check the ranking against the raw rows page by page.

It looks right, but checking by hand at this scale wasn't practical, and I couldn't easily trace ranked pages back to their raw rows.

So I'd treat the list as **a place to start a human review, not a verdict.**

> **Before publishing:** Add one line on how low-data pages are handled, pushed down or just labeled. And if you can, check the page that dropped 92% but scored 0. That might be a gap in the rule, and it's worth knowing.
