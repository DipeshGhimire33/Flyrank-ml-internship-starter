## 1. Two paper findings + my methodology questions

*Pick two findings from the FlyRank research paper. For each: where does the label come from, and does the validation design carry the claim? Constructive tone.*

I would frame the critique constructively: the findings may be useful, but the **label definition and validation design** determine how strongly they can support the conclusions.

### Finding 1 — Content refresh can improve performance

* **Where does the label come from?** The paper should clearly identify whether “successful refresh” is based on an observed post-refresh performance change, a manually assigned label, or another proxy.
* **Methodology question:** If the label is based on performance after a refresh, was there a comparable control group or counterfactual? Without one, improvement could also reflect seasonality, search-demand changes, or other external factors.
* **Assessment:** The validation supports an **association** with post-refresh performance only if the outcome and comparison period are clearly defined; it does not automatically establish causation.

### Finding 2 — Certain page signals identify strong refresh candidates

* **Where does the label come from?** The paper should explain whether candidate quality is defined by the same signals used as model features or by an independent post-refresh outcome.
* **Methodology question:** If the label is constructed from the same variables used for prediction, there is a risk of **target leakage or circular validation**.
* **Assessment:** Validation is more convincing when the target is independently measured and the evaluation data represent pages/time periods not used to construct the rule.

**Methodology takeaway:** The key questions are **“How was the label created?”** and **“Does the validation setup test performance on genuinely unseen information?”** These determine how much confidence we should place in the paper's findings.

## 3. Leakage audit

*The same hunt from Week 3, on your final feature set.*

```python
# Leakage audit on the final feature set

feature_cols = [
    "days_since_last_update",
    "impressions_90d",
    "clicks_90d",
    "impressions_last_30d",
    "clicks_last_30d",
    "impressions_prev_30d",
    "clicks_prev_30d",
    "search_volume",
    "avg_position",
    "ctr"
]

excluded = [
    "content_id",
    "client_id",
    "provider_used",
    "model_used",
    "trend_direction",
    "trend_pct"
]

future_like = [
    col for col in feature_cols
    if any(word in col.lower() for word in ["future", "next", "forecast"])
]

used_excluded = [
    col for col in feature_cols
    if col in excluded
]

print("Excluded fields accidentally used:", used_excluded)
print("Potential future-window features:", future_like)
```

The final feature set was checked for identifier fields, process fields, trend information, and future-looking variables. `trend_direction` and `trend_pct` were excluded because they contain trend information that could leak the outcome. No excluded or future-window features were used in the final feature set.
