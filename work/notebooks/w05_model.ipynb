## 4. Errors and interpretation

*Where is the model wrong? What does it lean on? A short error analysis beats a big metric table.*

The Opportunity Score can over-prioritize pages with temporary traffic declines, high visibility, or long periods since the last update, even when a refresh is not actually needed. It can also under-prioritize newer pages or pages with missing data because there is less evidence available.

The score mainly leans on recent performance change, content freshness, search visibility, average position, and CTR. These signals help identify pages worth reviewing, but they do not prove that refreshing the page will improve performance. Human review is still required before taking action.
