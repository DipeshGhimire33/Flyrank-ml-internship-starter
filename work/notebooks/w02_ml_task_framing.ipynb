## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A page's refresh opportunity depends on **multiple signals**, not just one condition.
For example, a page can be old but still performing well.
Another page can be recent but have a `down` trend.
Performance, freshness, engagement, and trend can point in different directions.
Because these signals interact, one simple **if-statement** may not rank pages well.
A scoring approach can combine these signals and give a more useful priority ranking.
